# Generated Code

```python
# To run this code you need to install the following dependencies:

import os, datetime
from google import genai
from google.colab import userdata, files

def generate():
    client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))

    # Switch models here
    model_old = "models/imagen-4.0-ultra-generate-preview-06-06"
    # model="models/imagen-4.0-generate-preview-06-06"
    model_norml = "imagen-4.0-generate-001"
    model_imagen_ultra = "imagen-4.0-ultra-generate-001"
    model_imagen_fast = "imagen-4.0-fast-generate-001"

    # Prompt 1
    prompt_1 = """YouTube thumbnail for a relaxing reggae music video mix: on the left side,
      featuring a realistic professional photorealistic photograph of two beautiful women with beach attire, dreadlocks and open shirt /* and flat belly visible */
      walking through a luscious jungle, no illustration and a title in reggae-insprired colours on the right side merging into a seamless picture together"""

    # Prompt 2
    prompt_2 = """YouTube thumbnail for a relaxing reggae music video mix featuring a very realistic
      professional photography of beautiful woman with beach attire, dreadlocks and open shirt and
      fit flat belly visible performing at the DJ decks in a neatly decorated hut in the jungle
      and a large title in reggae-insprired colours on the right side merging into a seamless picture together"""

    # Prompt 3
    prompt_3 = """YouTube thumbnail for a relaxing reggae music video mix: on the left side,
      A realistic professional photorealistic picture taken by Canon R5 captures a smiling young woman
      with fair skin and striking blue eyes, adorned with intricate dreadlocks and tribal jewelry.
      She's standing in a tropical beachside hut with a thatched roof, surrounded by blue ocean waves and white sand.
      The background features wooden furniture and a relaxed, bohemian atmosphere. The woman's radiant smile and
      colorful accessories enhance the scene's exotic, carefree vibe. On the right side a title in reggae-insprired colours on
      the right side merging into a seamless picture together without a split"""

    # Prompt 4
    prompt_4 = """YouTube thumbnail for a one hour video mix of phonk electronic music featuring two parts:
      a very realistic full-body photo shoot of two 26-year old women. They have delicate toned features,
      and are dressed in a futuristic outfit, bikers shorts, with glowing neon accents.
      On the right side a large engaging title "1-Hour Phonk Music Mix" in cyberpunk-insprired colours and futuristic style.
      Both parts merging into a seamless picture together.
      The style is highly detailed and realistic, with a cyberpunk neon aesthetic, electrostatic lighnings.
    """

    # Prompt 5
    prompt_5 = """Vibrant YouTube thumbnail for a one-hour video mix of summer electronic music featuring two parts: two people, a man and a woman,
      posing against a tropical beach backdrop with palm trees and a sports car. The man, with a muscular build, wears a colorful Hawaiian shirt, black shorts, and sunglasses.
      The woman, with long dark hair, is in a white crop top and high-waisted denim shorts. Both appear confident, the woman touching her hair. The background is a bright, sunny sky.
      On the right side, a large engaging title "Summer Vibes 2025" is prominently displayed largely in summer-colorful letters. Both parts merge into a seamless picture together.
    """
        # with long, flowing hair and large, round glasses that reflect vibrant neon lights

    result = client.models.generate_images(
        model = model_imagen_ultra,
        prompt = prompt_4,
        config = dict(
            number_of_images=1,
            output_mime_type="image/jpeg",
            person_generation="ALLOW_ADULT",
            aspect_ratio="16:9",
        ),
    )

    if not result.generated_images:
        print("No images generated.")
        return "NO"

    if len(result.generated_images) != 1:
        print("Number of images generated does not match the requested number.")

    for generated_image in result.generated_images:
        image = generated_image.image.show()

        timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
        filename = f"Google_Imagen_Colab_generated_image_{timestamp}.jpg"

        generated_image.image.save(filename)
        files.download(filename)
        print(f"Image saved as {filename}")

    return "YES"

if __name__ == "__main__":
    # Check if the GEMINI_API_KEY environment variable is set and not empty then retrieve it if necessary
    api_key = os.environ.get("GEMINI_API_KEY")
    if not api_key:
        print("GEMINI_API_KEY environment variable is not set or is empty. Retrieving...")
        os.environ["GEMINI_API_KEY"] = userdata.get("GOOGLE_API_KEY")
    else:
        print("API Key is set already")

    result = ""
    count = 1
    while result != "YES" and count < 5:
      count = count + 1;
      result = generate()
```
